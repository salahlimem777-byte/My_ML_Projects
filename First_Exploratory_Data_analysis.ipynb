# Step 2: Data Preperation
- Dropping irrelevant columns and rows
- Identifying duplicated columns
- Renaming Columns
- Feature Creation

```python
# Example of dropping columns
# df.drop(['Opening date'], axis=1)
```

```python
df=df[['coaster_name', 'Length', 'Speed',
    'Location',
    #'Status', 'Opening date',
       #'Type',
    'Manufacturer',
    #'Height restriction', 'Model', 'Height',
       #'Inversions', 'Lift/launch system', 'Cost', 'Trains', 'Park section',
       #'Duration', 'Capacity', 'G-force', 'Designer', 'Max vertical angle',
       #'Drop', 'Soft opening date', 'Fast Lane available', 'Replaced',
       #'Track layout', 'Fastrack available', 'Soft opening date.1',
       #'Closing date',
    #'Opened', 
    #'Replaced by', 'Website',
       #'Flash Pass Available', 'Must transfer from wheelchair', 'Theme',
       #'Single rider line available', 'Restraint Style',
       #'Flash Pass available', 'Acceleration', 'Restraints', 'Name',
       'year_introduced', 'latitude', 'longitude', 'Type_Main',
       'opening_date_clean',
    #'speed1', 'speed2', 'speed1_value', 'speed1_unit',
       'speed_mph', 
    #'height_value', 'height_unit', 
    'height_ft',
       'Inversions_clean', 'Gforce_clean']].copy()
```

```python
df["opening_date_clean"]=pd.to_datetime(df["opening_date_clean"])
```

```python
df.columns
```

```python
df=df.loc[~df.duplicated(subset=['coaster_name','Location','opening_date_clean'])]
```

```python
df.shape
```

```python
df.loc[df.duplicated(subset=['coaster_name'])].head()
```

```python
df=df.reset_index(drop=True).copy()
```

# Step 3: Feature Understanding

- Plotting Feature Distributions

```python
ax=df['year_introduced'].value_counts().head(10).plot(kind='bar',title='Top 10 years Coasters introduced')
ax.set_xlabel('Year introduced')
ax.set_ylabel('Number of Coasters')
```

# Step 4: Feature Relationships
- Scatterplot
- Heatmap Correlation
- Pairplot
- Groupby comparisons

```python
df.plot(kind='scatter',x='speed_mph',y='height_ft', title='Speed vs height')
plt.show()
```

```python
sns.scatterplot(x='speed_mph',y='height_ft',data=df,hue='year_introduced')
```

```python
sns.pairplot(df,vars=['year_introduced','speed_mph','height_ft','Inversions_clean','Gforce_clean'],hue='Type_Main')
plt.show()
```

```python
df_corr=df[['year_introduced','speed_mph','height_ft','Inversions_clean','Gforce_clean']].dropna().corr()
df_corr
```

# Step 5: Ask a Question about the data
- Try to answer a question you have about the data using a plot or statistic.

What are the locations with the fastest roller coasters (minimum of 10)?

```python
df.query('Location!= "Other"').groupby("Location")["speed_mph"].agg(['mean','count']).query('count >= 10').sort_values('mean').plot(kind='barh')
```
