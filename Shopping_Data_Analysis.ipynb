```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("/kaggle/input/datasets/iamsouravbanerjee/customer-shopping-trends-dataset/shopping_trends.csv")
```

```python
numeric_col=df.select_dtypes(include=['int64','float64']).columns
numeric_col
```

```python
numeric_col=['Purchase Amount (USD)', 'Review Rating','Previous Purchases']
df[numeric_col].hist(bins=30)
plt.suptitle('Purchase and Rating details')
plt.show()
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col].value_counts().plot(kind='bar')
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
def corolation_Matrix(df):
    numeric_df = df.select_dtypes(include=[np.number])
    corr_matrix = numeric_df.corr()
    plt.figure(figsize=(15, 10))
    sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm')
    plt.title('Correlation Matrix')
    plt.show()
```

```python
corolation_Matrix(df)
```
